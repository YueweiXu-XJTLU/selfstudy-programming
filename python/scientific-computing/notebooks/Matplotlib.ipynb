# Matplotlib

## Basic Manipulations

### Draw

```python
x = [1, 2, 3, 4, 5]
y = [1, 8, 27, 64, 125]
```

#### Matlab Way

```python
# plt.figure() + plt.plot()
fig1 = plt.figure() # create a figure object
plt.plot(x, y)
```

#### OOP Way

```python
# plt.figure() + plt.axes() + axes.plot()
fig2 = plt.figure()
ax2 = plt.axes() # the axes object is a rectangle with numbers on x & y axes
ax2.plot(x, y)
```

## Multiple Figures

> One Code Cell for One Figure

### Multiple Lines

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [1, 4, 6, 16, 25]
y3 = [1, 8, 27, 64, 125]
```

```python
fig1 = plt.figure()
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
```

### Multiple Sub-figures

#### Matlab Way

```python
# plt.subplot
fig1 = plt.figure()
plt.subplot(3, 1, 1), plt.plot(x, y1) # the final figure has 3 row(s) * 1 col(s), this figure is the 1st one
plt.subplot(3, 1, 2), plt.plot(x, y2) # the final figure has 3 row(s) * 1 col(s), this figure is the 2nd one
plt.subplot(3, 1, 3), plt.plot(x, y3) # the final figure has 3 row(s) * 1 col(s), this figure is the 3rd one
```

#### OOP Way

```python
# plt.subplots() & ax[].plot()
fig2, ax2 = plt.subplots(3) # the final figure has 3 sub-figures, if no number is given (`plt.subplots()`), draw 1 figure by default
ax2[0].plot(x, y1) # this figure is the 1st one
ax2[1].plot(x, y2) # this figure is the 2nd one
ax2[2].plot(x, y3) # this figure is the 3rd one
```

## Figure Attributes

### A Compositive Function of OOP
> Except these 5 attributes, different between Matlab Way & OOP Way is just `plt.xxx()` --> `ax.xxx()`

```python
# ax1.set(xlim=(), ylim=(), title='', xlabel='', ylabel='')
x = [1, 2, 3, 4, 5]
y = [1, 8, 27, 64, 125]
fig1, ax1 = plt.subplots()
ax1.plot(x, y)
ax1.set(xlim=(1, 10), ylim=(1, 200), title='Title', xlabel='X-axis', ylabel='Y-axis')
```

### Axes Ranges

```python
x = [1, 2, 3, 4, 5]
y = [1, 8, 27, 64, 125]
```

#### Use 'lim'

##### Matlab Way

```python
# plt.xlim() / plt.ylim()
fig1 = plt.figure()
plt.plot(x, y)
plt.xlim(1, 10)
plt.ylim(1, 200)
```

##### OOP Way

```python
# ax2.set_xlim() / ax2.set_ylim()
fig2, ax2 = plt.subplots()
ax2.plot(x, y)
ax2.set_xlim(1, 10)
ax2.set_ylim(1, 200)
```

#### Use 'axis'

```python
# plt.axis()
fig3 = plt.figure()
plt.plot(x, y)
plt.axis([1, 10, 1, 200]) # [x_lower, x_upper, y_lower, y_upper]
```

### Unit Length

```python
x = [1, 2, 3, 4, 5]
y = [1, 8, 27, 64, 125]
```

```python
# plt.axis('equal')
# make unit length of x & y axes become SAME
fig1 = plt.figure()
plt.plot(x, y)
plt.axis('equal')
```

### Title & Axes Labels

```python
x = [1, 2, 3, 4, 5]
y = [1, 8, 27, 64, 125]
```

#### Matlab Way

```python
# plt.title & plt.xlabel / plt.ylabel
fig1 = plt.figure()
plt.plot(x, y)
plt.title('Title')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

#### OOP Way

```python
# ax.set_title & ax.set_xlabel / ax.set_ylabel
fig2, ax2 = plt.subplots()
ax2.plot(x, y)
ax2.set_title('Title')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')
```

### Legend

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [0, 0, 0, 0, 0]
y3 = [-1, -2, -3, -4, -5]
```

```python
# plt.plot(x, y, label='...') & plt.legend()
fig1 = plt.figure()
plt.plot(x, y1, label='y1 = x')
plt.plot(x, y2, label='y2 = 0')
plt.plot(x, y3, label='y3 = -x')
plt.legend(loc='best', frameon=True, ncol=1)
# plt.legend(loc='center', frameon=False, ncol=3)
# `loc='upper/center/lower left/center/right'/'center'/'best'`: location
# `frameon=True`: frame of legend is on
# `ncol=1`: number of column of the legend
```
