# [PyTorch](https://pytorch.org/get-started/locally/)

```python
import torch
print(torch.__version__)

# specify data type
dtype = torch.float32

# specify calculation device
print(torch.cuda.is_available())
device = torch.device("cuda:0") # open terminal --> `nvidia-smi` --> `0  NVIDIA GeForce RTX 4090 ...  WDDM`
# device = torch.device("cpu")
# device = torch.device("cuda: 0" if torch.cuda.is_available() else "cpu")
# device = torch.device("mps") # for MAC (Apple Silicon)
```

## Basic PyTorch

### Create Tensors

```python
# create 2 random tensor
ts1 = torch.randn(2, 3, device=device, dtype=dtype)
ts2 = torch.randn(2, 3, device=device, dtype=dtype)
print(ts1)
print(ts2)
# first convert into tensor in CPU, then convert into numpy array
# ts1.cpu()
# ts2.cpu()
```

### Basic Functions

```python
# multiply by element
print("multiply by element:", ts1 * ts2)

# sum up all elements
print("sum up all elements:", ts1.sum())

# get element by index
print("get element by index:", ts1[1, 2]) # [row, col], both start from 0

# get max-valued element
print("get max-valued element:", ts1.max())
```
