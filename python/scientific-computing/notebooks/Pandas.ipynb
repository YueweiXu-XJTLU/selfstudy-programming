## Objects

### 1-D Object - Create

```python
# pd.Series(dict)
# use dict: {index:value}
sr =  pd.Series({'a':1, 'b':2, 'c':3})
print(sr)
sr
```

### 2-D Object - Attributes

```python
sr_age = pd.Series({'1号':53, '2号':64, '3号':72, '4号':82})
sr_gender = pd.Series({'1号':'女', '2号':'男', '3号':'男', '4号':'女'})
df = pd.DataFrame({'年龄':sr_age, '性别':sr_gender})

# df.values
print(df.values, type(df.values))
# use this to extract original array
# AND
# extract certain column:
arr_age = df.values[:,0].astype(int) # remember to cast into int/float
print(arr_age, type(arr_age))

# df.index
print(df.index, type(df.index))

# df.columns
print(df.columns, type(df.columns))
```

## Index

### 1-D Index

```python
sr =  pd.Series({1:'a', 2:'b', 3:'c'})
print(sr)
```

```python
# Explicit Index: use row/col names: sr.loc[]
# Element
print(sr.loc[1])

# Slice
print(sr.loc[1:3]) # label[1] - label[3]
```

```python
# Implicit Index: use implicit index 0, 1, ...: sr.iloc[]
# Element
print(sr.iloc[1])

# Slice
print(sr.iloc[1:3]) # index[1] - index[2]
```

### 2-D Index

```python
sr_age = pd.Series({'1号':53, '2号':64, '3号':72, '4号':82})
sr_gender = pd.Series({'1号':'女', '2号':'男', '3号':'男', '4号':'女'})
df = pd.DataFrame({'年龄':sr_age, '性别':sr_gender})
print(df)
```

```python
# Explicit Index: use row/col names: sr.loc[]
# Element
print(df.loc['1号', '年龄'])

# Slice
print(df.loc['2号':'3号', '年龄':'性别']) # label['2号'] - label['3号']; label['年龄'] - label['性别']
print('==========')
# allow:
print(df.loc['1号']) # omit ['1号', :] for a row
print('==========')
print(df['年龄']) # col is key for df
```

```python
# Implicit Index: use implicit index 0, 1, ...: sr.iloc[]
# Element
print(df.iloc[0, 0])

# Slice
print(df.iloc[1:3, :]) # index[1] - index[2]; index[start] - index[end]
```

```python
# Fancy Index: return a matrix with row/col-labels
print(df.loc[['1号', '3号'], ['年龄', '性别']])
print(df.iloc[[0, 1], [1, 0]])
```

### Concatenate

```python
# pd.concat()
# 1-D
k1 = ['no.1', 'no.2', 'no.3', 'no.4']
v1 = [1, 2, 3, 4]
sr_1 = pd.Series(v1, index=k1)

k2 = ['no.4', 'no.5', 'no.6']
v2 = [4, 5, 6]
sr_2 = pd.Series(v2, index=k2)

sr_all = pd.concat([sr_1, sr_2])
print(sr_all)
print(sr_all.index.is_unique)

# 2-D
v1 = [[10, '女'], [20, '男'], [30, '男'], [40, '女']]
v2 = [[1, '是'], [2, '是'], [3, '是'], [4, '否']]
v3 = [[50, '男', 5, '是'], [60, '女', 6, '是']]
i1 = ['1号', '2号', '3号', '4号']
i2 = ['1号', '2号', '3号', '4号']
i3 = ['5号', '6号']
c1 = ['年龄', '性别']
c2 = ['牌照', 'ikun']
c3 = ['年龄', '性别', '牌照', 'ikun']
df1 = pd.DataFrame(v1, index=i1, columns=c1)
df2 = pd.DataFrame(v2, index=i2, columns=c2)
df3 = pd.DataFrame(v3, index=i3, columns=c3)
print(df1)
print(df2)
print(df3)
print('==========')

df_by_col = pd.concat([df1, df2], axis=1) # add columns
print(df_by_col)
print('==========')
df_by_row = pd.concat([df_by_col, df3]) # default axis=0 # add rows
print(df_by_row)
```

## Calculations

## Missing Values

### Discover Missing Values

```python
# df.isnull() / df.notnull()
v = [[10, 'Female'], [None, 'Male'], [30, None], [40, 'Female']]
i = ['No.1', 'No.2', 'No.3', 'No.4']
c = ['Age', 'Gender']
df = pd.DataFrame(v, index=i, columns=c)
print(df)
print('==========')

print(df.isnull())
print('==========')
print(df.notnull()) # same as `~df.isnull()`
```

### Delete Missing Values

```python
# df.dropna()
v = [[10, 'Female'], [None, 'Male'], [None, None], [40, 'Female']]
i = ['No.1', 'No.2', 'No.3', 'No.4']
c = ['Age', 'Gender']
df = pd.DataFrame(v, index=i, columns=c)
print(df)
print('==========')

print(df.dropna()) # default `axis=0` & `how='any'`
print('==========')
print(df.dropna(axis=1))
print('==========')
print(df.dropna(how='all')) # only delete when all NaN
print('==========')
print(df.dropna(thresh=1)) # only delete when #NaN > thresh
print('==========')
```

### Fill In Missing Values

```python
# df.fillna()
v = [[10, 'Female'], [None, 'Male'], [None, None], [40, 'Female']]
i = ['No.1', 'No.2', 'No.3', 'No.4']
c = ['Age', 'Gender']
df = pd.DataFrame(v, index=i, columns=c)
print(df)
print('==========')

# fill in a number
print(df.fillna(0))
print('==========')

# fill in the forward number
# we can add `axis=1`, but it usually does not make sense
# print(df.fillna(method='ffill'))
# OR
print(df.ffill())
print('==========')

# fill in the backward number
# we can add `axis=1`, but it usually does not make sense
# print(df.fillna(method='bfill'))
# OR
print(df.bfill())
print('==========')
```

### Pivot Table
> 数据透视

```python
# df.pivot_table()
# calculate .mean() by default, but we can point out `aggfunc='max'` / ...
df = pd.read_csv('Data.csv', index_col=0)
print(df)
print('==========')

# 1 feature
print(df.pivot_table('Ikun', index='Gender')) # how values of `Gender` affect value of `Ikun`
print('==========')

# 2 features
print(df.pivot_table('Ikun', index='Gender', columns='Age')) # how values of `Gender` & 'Age' affect value of `Ikun`
print('==========')

# multiple features
# pd.cut() & pd.qcut(): continuous data --classify--> discrete data
Num1 = pd.cut(df['Num'], [0, 10, 20]) # classify data in df['Num'] into (0, 10] & (10, 20]
Num2= pd.cut(df['Num'], 2) # classify data in df['Num'] into 2 intervals automatically
print(df.pivot_table('Ikun', index=['Gender', Num1], columns='Age', observed=True)) # `observed=True`: only output combination of index/columns that are observed
```
